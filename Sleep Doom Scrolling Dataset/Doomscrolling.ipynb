## Handling Missing Values

There are 6 features with missing values. The missing values need to be fixed using one of the imputation method namely, mean, median or mode.

```python
df[df['caffeine_intake_mg_per_day'].isna()].head()
```

Missing values in the column 'caffine_intake_mg_per_day' can be replaced by imputing the mean(average) value of the column.
Let us fix it.

```python
df['caffeine_intake_mg_per_day'] = df['caffeine_intake_mg_per_day'].fillna((df['caffeine_intake_mg_per_day']).mean())
```


###sleep_quality_score
*   Sleep quality score is an important feature, imputing it could result in unreliable data
*   Deleting is a good option here.



```python
df['sleep_quality_score'].value_counts()
```

```python
(df['sleep_quality_score']).mode()
```

```python
df = df.dropna(subset=['sleep_quality_score'])
```

### exercise_minutes_per_day

Imputing 'exercise_minutes_per_day' with mean

```python
df['exercise_minutes_per_day'] = df['exercise_minutes_per_day'].fillna(df['exercise_minutes_per_day'].mean())
```

```python
df=df.drop('respondent_id', axis=1)
```

```python
dumm_df = pd.get_dummies(df, columns=['gender','occupation_status','country_region','primary_device_used_at_night', 'bedtime_routine_type', 'uses_night_mode', 'keeps_phone_in_bedroom', 'consumes_negative_news_content', 'uses_sleep_tracking_app'],drop_first=True, dtype='int')
```

```python
x = dumm_df.drop(['doomscroller','sleep_quality_category'], axis=1)
y_doom = dumm_df['doomscroller']
y_qual = dumm_df['sleep_quality_category']
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaled_x = scaler.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split


X_train, X_test, Y_train, Y_test = train_test_split(scaled_x, y_doom, test_size=0.2, random_state=42, )


Y_train
```

###Decision Tree classifier

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()

model.fit(X_train, Y_train)
```

```python
pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy = accuracy_score(Y_test, pred)

print('Accuracy:', accuracy)
```

###Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier

model2 = RandomForestClassifier()

model2.fit(X_train, Y_train)

pred = model2.predict(X_test)

accuracy2 = accuracy_score(Y_test, pred)

print('Accuracy:', accuracy2)
```

###Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

model3 = LogisticRegression()

model3.fit(X_train, Y_train)

pred = model3.predict(X_test)

accuracy3 = accuracy_score(Y_test, pred)

print('Accuracy:', accuracy3)
print(classification_report(Y_test, pred))
```

###Support Vector Machines

```python
from sklearn.svm import SVC

model4 = SVC()

model4.fit(X_train, Y_train)

pred = model4.predict(X_test)

accuracy4 = accuracy_score(Y_test, pred)

print('Accuracy:', accuracy4)
print(classification_report(Y_test, pred))
```

###K-Nearest Neighbours

```python
from sklearn.neighbors import KNeighborsClassifier

model5 = KNeighborsClassifier()

model5.fit(X_train, Y_train)

pred = model5.predict(X_test)

accuracy5 = accuracy_score(Y_test, pred)

print('Accuracy:', accuracy5)
```

###Naive Bayes Classifier

```python
from sklearn.naive_bayes import GaussianNB

model6 = GaussianNB()

model6.fit(X_train, Y_train)

pred = model6.predict(X_test)

accuracy6 = accuracy_score(Y_test, pred)

print('Accuracy:', accuracy6)
```

###Gradient Boost Classifier


```python
from sklearn.ensemble import GradientBoostingClassifier

model7 = GradientBoostingClassifier(n_estimators=50, max_depth=3, learning_rate=0.05, random_state=42)

model7.fit(X_train, Y_train)

pred = model7.predict(X_test)

accuracy7 = accuracy_score(Y_test, pred)

print('Accuracy:', accuracy7)
print(classification_report(Y_test, pred))
```

###XGBoost Classifier

```python
import xgboost as xgb

model8 = xgb.XGBClassifier(n_estimators=50, max_depth=1, learning_rate=0.05, random_state=42)

Y_train_int = Y_train.map({'Yes':1,'No':0})
Y_test_int = Y_test.map({'Yes':1,'No':0})

model8.fit(X_train, Y_train_int)

pred8 = model8.predict(X_test)

accuracy8 = accuracy_score(Y_test_int, pred8)

print('Accuracy:', accuracy8)
print(classification_report(Y_test_int, pred8))
```
